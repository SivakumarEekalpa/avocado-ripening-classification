# 02 — Classical Computer Vision + Machine Learning
Train and compare tuned SVM and Random Forest using color histograms, LAB statistics, LBP and GLCM descriptors.

```python
from pathlib import Path
import sys, os, json, time
import numpy as np, pandas as pd
import matplotlib.pyplot as plt
ROOT = Path.cwd()
if not (ROOT/"src").exists() and (ROOT.parent/"src").exists(): ROOT=ROOT.parent
if str(ROOT) not in sys.path: sys.path.insert(0,str(ROOT))
DATA_DIR = ROOT/"data"/"raw"
OUTPUT_DIR = ROOT/"outputs"
for d in ["plots","confusion_matrices","metrics","predictions"]: (OUTPUT_DIR/d).mkdir(parents=True,exist_ok=True)
print("Project root:", ROOT, "| Dataset:", DATA_DIR)
```

```python
from src.dataset import discover_dataset, stratified_split, save_split_manifest
from src.classical_model import train_classical
from src.utils import set_seed
set_seed(42)
df=discover_dataset(DATA_DIR)
train_df,val_df,test_df=stratified_split(df)
save_split_manifest(train_df,val_df,test_df,ROOT/"data"/"processed"/"split_manifest.csv")
print(len(train_df),len(val_df),len(test_df))
display(pd.concat([train_df.label.value_counts(normalize=True).rename("train"),
                   val_df.label.value_counts(normalize=True).rename("validation"),
                   test_df.label.value_counts(normalize=True).rename("test")],axis=1).fillna(0))
```

```python
results=train_classical(train_df,val_df,test_df,
                        output_dir=ROOT/"models",metrics_dir=OUTPUT_DIR/"metrics")
display(results[["model","accuracy","precision_macro","recall_macro","f1_macro","training_seconds","best_params"]])
```
